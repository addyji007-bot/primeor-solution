# Remove ONLY exact duplicates

**NOTE**: one row = one product line within an order.
order_id is expected to repeat. It is NOT a unique row identifier.

The only duplicate you can safely remove without investigation is a row that is identical across every single column. Anything else needs manual inspection first

```python
exact_dupes = df.duplicated().sum()
print("Exact duplicate rows:", exact_dupes)

df = df.drop_duplicates()

print("New shapes:", df.shape)
print("Duplicates remaining:", df.duplicated().sum())
```

```python
df['customer_name'] = df['customer_name'].fillna("Unknown Custommer")
```

```python
df[df['profit'].isnull()]
```

```python
print(df['sales'].dtype)
print(df['profit'].dtype)
```

```python
# remove commas and convert to proper number
df['sales'] = df['sales'].astype(str).str.replace(',', '', regex=False)
df['sales'] = pd.to_numeric(df['sales'], errors='coerce')

# do the same check for other numeric columns just in case
for col in ['quantity', 'discount', 'profit', 'shipping_cost']:
    print(col, df[col].dtype)
```

```python
df['profit_estimated'] = df['profit'].isnull()  # flag BEFORE filling

margin_by_subcat = df.dropna(subset=['profit']).groupby('sub_category').apply(
    lambda x: x['profit'].sum() / x['sales'].sum()
)

def fill_profit(row):
    if pd.isnull(row['profit']):
        return row['sales'] * margin_by_subcat.get(row['sub_category'], 0)
    return row['profit']

df['profit'] = df.apply(fill_profit, axis=1)
```

```python
df.isnull().sum()
```

# Cleaning text col

```python
text_cols = ['ship_mode', 'segment', 'state', 'country', 'market', 
             'region', 'category', 'sub_category', 'order_priority']

for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.title()
```

# Fixing Dates

```python
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce', dayfirst=True)
df['ship_date'] = pd.to_datetime(df['ship_date'], errors='coerce', dayfirst=True)

print(df['order_date'].isnull().sum())
print(df['ship_date'].isnull().sum())
```

```python
# logic check: ship_date should never be before order_date
bad_dates = df[df['ship_date'] < df['order_date']]
print("Rows where ship_date is before order_date:", len(bad_dates))
bad_dates.head()
```

```python
from pathlib import Path

def print_tree(path, max_depth=1, prefix="", depth=0):
    path = Path(path)
    items = sorted(path.iterdir(), key=lambda p: (p.is_file(), p.name.lower()))

    for i, item in enumerate(items):
        connector = "└── " if i == len(items) - 1 else "├── "
        print(prefix + connector + item.name)

        if item.is_dir() and depth < max_depth:
            extension = "    " if i == len(items) - 1 else "│   "
            print_tree(item, max_depth, prefix + extension, depth + 1)

# Change "." to your project path if needed
print(Path(".").resolve().name)
print_tree(".", max_depth=1)
```
